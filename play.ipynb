```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# load the images:
image0 = cv2.imread('debug_camera_image0.png')
image1 = cv2.imread('debug_camera_image1.png')
imdiff = cv2.imread('diff_image.png')
```

```python
plt.imshow(imdiff)
```

```python
plt.imshow(image1)
```

```python
gray0 = cv2.cvtColor(image0, cv2.COLOR_BGR2GRAY)
gray1 = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)

diff = gray0.astype(np.int32) - gray1.astype(np.int32)
diff = image1[:,:,0].astype(np.int32) - image0[:,:,0].astype(np.int32)

diff = np.clip(diff, 0, 255).astype(np.uint8)
plt.imshow(diff, cmap='gray');
```
